````python
## 🔎 Exercise 3 — Solutions


# 1. Select all orders where the sales are greater than 500
df[df["sales"] > 500]


# 2. Select all orders where the profit is negative
df[df["profit"] < 0]


# 3. Select all orders from the Technology category
df[df["category"] == "Technology"]


# 4. Select all orders where the sales are between 100 and 500
df[df["sales"].between(100, 500)]


# 5. Select all orders from the Furniture or Technology categories
df[df["category"].isin(["Furniture", "Technology"])]


# 6. Check the data type of the order_date column
df["order_date"].dtype


# 7. Create a new datetime column
df["order_date_formatted"] = pd.to_datetime(df["order_date"])


# 8. Create year, month, and day columns using the .dt accessor
df["order_year"] = df["order_date_formatted"].dt.year
df["order_month"] = df["order_date_formatted"].dt.month
df["order_day"] = df["order_date_formatted"].dt.day


# 9. Select all orders placed during 2019
df[df["order_date_formatted"].dt.year == 2019]


# 10. Select all orders placed between January 1, 2019 and June 30, 2019
df[
    df["order_date_formatted"].between(
        "2019-01-01",
        "2019-06-30"
    )
]


# 11. Select all orders placed during January, regardless of the year
df[df["order_date_formatted"].dt.month == 1]


# 12. Select all orders where product_name contains "Chair"
df[df["product_name"].str.contains("Chair")]


# 13. Select all orders where product_name starts with "A"
df[df["product_name"].str.startswith("A")]


# 14. Select all orders where product_name ends with "Table"
df[df["product_name"].str.endswith("Table")]


# 15. Create a column containing the number of characters in product_name
df["product_name_length"] = df["product_name"].str.len()


# 16. Select products containing "Office", ignoring uppercase/lowercase
df[df["product_name"].str.contains("office", case=False)]


# 17. Technology orders during 2019 with sales greater than 500
# Use the Mask technique
mask = (
    (df["category"] == "Technology") &
    (df["order_date_formatted"].dt.year == 2019) &
    (df["sales"] > 500)
)

df[mask]


# 18. Products containing "Chair" with positive profit
df[
    (df["product_name"].str.contains("Chair")) &
    (df["profit"] > 0)
][["product_name", "sales", "profit"]]


# 19. West region orders during January 2019
# Use the Mask technique
mask = (
    (df["region"] == "West") &
    (df["order_date_formatted"].dt.year == 2019) &
    (df["order_date_formatted"].dt.month == 1)
)

df[mask][["order_date_formatted", "customer", "sales"]]


# 20. Products containing "Table" or "Chair" with sales greater than 200
df[
    (df["product_name"].str.contains("Table|Chair")) &
    (df["sales"] > 200)
]


# 21. 2019 orders containing "Office", sales > 300, and positive profit
# Use the Mask technique
mask = (
    (df["order_date_formatted"].dt.year == 2019) &
    (df["product_name"].str.contains("Office", case=False)) &
    (df["sales"] > 300) &
    (df["profit"] > 0)
)

df[mask][
    ["order_date_formatted", "customer", "product_name", "sales", "profit"]
].sort_values("profit", ascending=False)


# 22. Create Customer First Name and find names starting with "M"
df["customer_first_name"] = df["customer"].str.split().str[0]

df[df["customer_first_name"].str.startswith("M")].head()

````

##4

````python
## 🧹 Exercise 4 — Solutions


# 1. Check how many missing values exist in each column
df.isna().sum()


# 2. Select all rows where the zip value is missing
df[df["zip"].isna()]


# 3. Fill the missing values in the zip column with 0
df["zip"] = df["zip"].fillna(0)


# 4. Convert the zip column to an integer data type
df["zip"] = df["zip"].astype(int)


# 5. Rename the manufactory column to manufacturer
df = df.rename(columns={"manufactory": "manufacturer"})


# 6. Create three sales levels: Low, Medium, High
df["sales_level"] = pd.cut(
    df["sales"],
    bins=3,
    labels=["Low", "Medium", "High"]
)


# 7. Divide profit into four quartiles
df["profit_quartile"] = pd.qcut(
    df["profit"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)


# 8. Calculate profit per item
df["profit_per_item"] = df["profit"] / df["quantity"]


# 9. Create a Boolean column for high-value orders
df["high_value_order"] = df["sales"] > 500


# 10. Create order size categories based on quantity
df["order_size"] = pd.cut(
    df["quantity"],
    bins=[0, 2, 5, float("inf")],
    labels=["Small", "Medium", "Large"]
)

````
