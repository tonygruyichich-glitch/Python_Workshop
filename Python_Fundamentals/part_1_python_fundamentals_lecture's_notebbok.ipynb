### `range()`

### Dictionaries

A dictionary is an **ordered and changeable collection of key-value pairs**.

Unlike a list, a dictionary does not access elements by their position. Instead, values are accessed using their **keys**.

```python
person = {
    "name": "John",
    "age": 30,
    "city": "Tel Aviv"
}
```

Here, "name", "age", and "city" are the keys, and "John", 30, and "Tel Aviv" are their corresponding values.

### Important Facts

- A dict is an ordered, mutable collection of key-value pairs.
- Dictionaries store data as key → value relationships.
- Each key must be unique.
- Values can be duplicated.
- Values can contain different data types.
- Values can be lists, sets, tuples, dictionaries, or other objects.
- Keys must be hashable, such as `str`, `int`, or `tuple`.
- Dictionaries are mutable, so items can be added, changed, or removed.
- Dictionaries preserve insertion order.
- Dictionaries are accessed using keys, not numeric indexes.
- Dictionaries are created using curly brackets: `{}`.
- The `len()` function returns the number of key-value pairs.
- Accessing a key that does not exist raises a `KeyError`.
- `get()` can be used to safely access a key without raising a `KeyError`.
- Common methods include:
  - `keys()` - returns all keys
  - `values()` - returns all values
  - `items()` - returns all key-value pairs
  - `get()` - retrieves a value safely
  - `update()` - adds or updates key-value pairs
  - `pop()` - removes a key-value pair
- A dictionary can contain another dictionary as a value, allowing nested dictionaries.
- A dictionary stores references to objects, rather than the actual objects themselves.

# ⚡ Data Retrieval Efficiency in Python: List, Dictionary & Set

When working with Python data structures, retrieval efficiency comes directly from how data is stored and accessed in memory under the hood.

---

## 1. List – Retrieval by Index: $O(1)$
Elements in a list are stored in contiguous memory locations. When you request an item by index (e.g., `lst[500]`), Python doesn't scan through items one by one. Instead, it calculates the exact memory address instantly using a simple formula:

$$\text{Memory Address} = \text{Base Address} + (\text{Index} \times \text{Element Size})$$

* **Result:** Retrieval happens in constant time ($O(1)$) — fetching an element at index `0` takes virtually the exact same time as fetching an element at index `1,000,000`.
* **Important Distinction (Searching by Value):** Checking if a value exists in a list (`x in lst`) is slow — **$O(n)$** — because Python must scan elements sequentially from the beginning.

---

## 2. Dictionary & Set – Retrieval by Key/Element: $O(1)$
Both Dictionaries and Sets are built on **Hash Tables**.
When you insert a key or search for an element, Python passes it through a **Hash Function**. This function instantly translates the data into a specific memory address.

* **Result:** Membership checks in a set (`x in my_set`) or dictionary lookups (`my_dict['key']`) happen in constant time **$O(1)$**.
* The system never scans through the data. It hashes the value and jumps directly to its dedicated slot in memory — whether the structure contains 10 items or 10 million.

---

## 📊 Comparison Summary

| Data Structure | Operation Type | Code Example | Time Complexity | How it Works in Memory |
| :--- | :--- | :--- | :---: | :--- |
| **List** | Retrieval by Index | `item = lst[5]` | **$O(1)$** | Direct mathematical calculation of memory address |
| **List** | Search by Value | `if x in lst:` | **$O(n)$** | Linear scan element-by-element from the start |
| **Dictionary** | Retrieval by Key | `val = d['key']` | **$O(1)$** | Maps key to memory address via Hash Function |
| **Set** | Membership Check | `if x in s:` | **$O(1)$** | Maps value to memory address via Hash Function |

---



````python
## 🧪 Live Proof Code (Run in a Colab Code Cell)

import time

# Create data structures with 1,000,000 items
size = 1_000_000
large_list = list(range(size))
large_set = set(range(size))

target = 999_999  # Item placed at the very end

# 1. Searching in a List - O(n)
start = time.perf_counter()
is_in_list = target in large_list
list_time = time.perf_counter() - start

# 2. Searching in a Set - O(1)
start = time.perf_counter()
is_in_set = target in large_set
set_time = time.perf_counter() - start

print(f"List search O(n): {list_time:.6f} seconds")
print(f"Set search  O(1): {set_time:.6f} seconds")
print(f"Set is ~{int(list_time / set_time):,} times faster!")
````

**What is a Hash Function?**
A mathematical "translation machine" that takes any input (text, number, or object) and converts it into a fixed integer identifier instantly.
* **Deterministic:** The input `"Bob"` will always produce the exact same integer during a program's execution.
* **Constant Speed:** Calculating a hash takes a fraction of a microsecond regardless of input size.

---

**How a Hash Table Works in Memory (RAM)**
When you create a Dictionary or Set, Python allocates a block of contiguous slots in system memory (**RAM**).

1. **Hash the Key:** Python computes `hash('Bob')` $\rightarrow$ `8,493,029,102`.
2. **Calculate the Index:** It maps this large number into the allocated memory range using a fast modulo operation:
   $$\text{Slot Index} = \text{Hash Value} \pmod{\text{Table Size}}$$
   For an 8-slot table: $8,493,029,102 \pmod 8 = \mathbf{6}$.
3. **Direct Memory Jump ($O(1)$):** Python jumps straight to slot **6** in RAM to store or fetch the data without searching through other items.

---

**Real-World Analogy: Automated Coat Check**
* **List Search $O(n)$:** A cloakroom attendant inspecting every single coat hanger one by one until finding your jacket.
* **Hash Table $O(1)$:** Entering your name into a kiosk, getting printed ticket `#42`, and walking directly to hanger 42.

## 🔀 Decision Making: `if`, `elif`, and `else`

Conditional statements allow your program to make decisions and run specific code blocks based on whether a condition evaluates to `True` or `False`.

* **`if`**: The initial check. Python evaluates this first; if it's `True`, its code block runs, and Python skips the rest.
* **`elif`** *(short for "else if")*: Tested **only if** all previous `if` or `elif` conditions were `False`. You can chain multiple `elif` blocks.
* **`else`**: The default fallback. Runs **only if** none of the preceding conditions evaluated to `True`.

---

### 🧪 Code Example

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")  # Executed because score is >= 80 and the first condition was False
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: Needs Improvement")

### 🐍 Python List Comprehensions: Elegance in a Single Line

**What is it?**  
A **List Comprehension** is a concise, highly readable, and Pythonic way to construct new lists from existing iterables—condensing multi-line `for` loops and `.append()` calls into a single line.

---

#### 💡 Comparison: Traditional Loop vs. List Comprehension

Creating a list containing the squares of even numbers from 1 to 10:

**Traditional Approach (4 lines):**
```python
squares = []
for x in range(1, 11):
    if x % 2 == 0:
        squares.append(x**2)

print(squares)  # [4, 16, 36, 64, 100]
```

#### List Comprehension (1 line):

```python
squares = [x**2 for x in range(1, 11) if x % 2 == 0]

print(squares)  # [4, 16, 36, 64, 100]



````python
"""Problem Statement:

Given a list of words, transform each string based on its length:

If a word has more than 4 characters, convert it to uppercase and append an exclamation mark ("!").

Otherwise, reverse the word."""

words = ["python", "is", "awesome", "code", "ai"]

# 1. Traditional Approach (7 lines)
transformed = []
for word in words:
    if len(word) > 4:
        transformed.append(word.upper() + "!")
    else:
        transformed.append(word[::-1])

print('1. ',transformed)

# 2. List Comprehension
[word.upper() + "!" if len(word) > 4 else word[::-1] for word in words]
````

### 4. Functions
**Concept:** Creating reusable blocks of code.

**SQL Connection:**
Think of a function as a **Stored Procedure** or a **User Defined Function (UDF)**. It allows you to write logic once and apply it to many different values.

**✍️ Practice:** Create a function that calculates a 10% tax on an amount and returns the total.

| Function Type | Description | Primary Use Case / Example |
| :--- | :--- | :--- |
| **User-Defined (UDF)** | Custom functions created with `def` | Multi-step logic: `def calc(x): return x * 2` |
| **Lambda Functions** | A lambda function is an anonymous, single-line function built for quick, inline data transformations | Quick inline operations in Pandas `.apply()` |
| **Built-in Functions** | Pre-packaged core Python functions | `len()`, `sum()`, `max()`, `min()`, `type()` |
| **Object Methods** | An object method is a function tied to a specific data type and called directly on it using dot notation | `"text".upper()`, `my_list.append(5)` |
| **Higher-Order** | Functions taking other functions as inputs | `map(function, iterable)` |

````python
# --- Example: UDF ---
def greet_user(name):
    return f"Hello, {name}! Welcome to the data platform."

# Using the function
message = greet_user("Dana")
print(message)
print(greet_user("Leon"))
````

````python
#high order function using UDF
li = [4,7,9]
def mul(x):
  return x*2
print(list(map(mul, li)))
````

````python
#high orderfunction using lambda function
li = [4,7,9]
print(list(map(lambda x: x*2, li)))
````

````python
#UDF using core function
li = [4,7,9]
def sum_plus_1(x:list):
  return sum(x)+1
print(sum_plus_1(li))
````
